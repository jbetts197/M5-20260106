First bring in the python libaries

```python
import pandas as pd
from datetime import datetime
```

**Cleanse checklist:**
- Brutal approach
1. Remove any empty rows
2. Standardize text data in customer names (trim whitespace)

Prereq step, import the data to a dataframe:

```python
input_df = pd.read_csv('../raw_data/03_Library SystemCustomers.csv')
print(input_df)
```

Step 1: Remove any empty and duplicate rows

```python
result_df = input_df.dropna(how='all').drop_duplicates()
print(result_df)
```

Step 2. Remove whitespaces from customers name

```python
result_df['Customer Name'].str.strip()
print(result_df)
```

Finally export all the records to output csv

```python
# Remove the valid_record column and export
result_df.to_csv("../output_cleansed_data/cleansed_system_customers.csv", index=False)
```
