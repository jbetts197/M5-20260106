**Cleanse checklist:**
- Brutal approach
- Create columns: valid_checkout, valid_returned, valid_record columns
- When final dataframe export, only export records where all set to true
1. Remove any empty rows
2. Remove any duplicate rows
3. Handle NAN values
4. Fix date formatting issues
5. Standardize text data in book titles (trim whitespace)

Prereq step, import the data to a dataframe:

```python
input_df = pd.read_csv('../raw_data/03_Library Systembook.csv')
print(input_df)
```

Step 1 and Step 2: Remove any empty and duplicate rows

```python
result_df = input_df.dropna(how='all').drop_duplicates()
print(result_df)
```

Step 3. Handle any NaN values

```python
# If Customer ID is NaN, drop record as not useful
result_df['valid_record'] = result_df['Customer ID'].notna()
print(result_df)
```

Step 4. Handle date formatting issues

```python
# Remove double quotes from columns
result_df['Book checkout'] = result_df['Book checkout'].str.replace('"', '')
result_df['Book Returned'] = result_df['Book Returned'].str.replace('"', '')

# Define regex pattern for DD/MM/YYYY format
date_pattern = r'^\d{2}/\d{2}/\d{4}$'
result_df['valid_record'] = result_df['Book checkout'].str.match(date_pattern, na=False)
result_df['valid_record'] = result_df['Book Returned'].str.match(date_pattern, na=False)
print(result_df)
```
